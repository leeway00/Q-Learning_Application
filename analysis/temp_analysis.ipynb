# Temp Analysis

Quick sanity checks for AI Clinician core outputs.

```python
import numpy as np
from pathlib import Path

from config import MDP_OUTPUT_DIR

# Update these paths as needed
npz_path = None
for p in sorted(MDP_OUTPUT_DIR.glob('mdp_core_*.npz')):
    npz_path = p

if npz_path is None:
    raise FileNotFoundError('No mdp_core_*.npz found in data/mdp_output')

print('Using:', npz_path)
data = np.load(npz_path, allow_pickle=True)
recqvi = data['recqvi']
OA = data['OA']
idxs = data['idxs']
print('recqvi shape:', recqvi.shape)
print('OA shape:', OA.shape)
print('idxs shape:', idxs.shape)
```

```python
cols = [0,3,4,5,6,18,19,20,21,22,23]
print(np.isnan(recqvi[:, cols]).mean(axis=0))
print(np.nanmean(recqvi[:, cols], axis=0))
```

```python
# Basic sanity checks
print('recqvi nan fraction:', np.isnan(recqvi).mean())
print('OA nan fraction:', np.isnan(OA).mean())
print('idxs nan fraction:', np.isnan(idxs).mean())

# Key metrics (columns based on Python fill):
# 0: model id
# 3: mean bootql (train)
# 5: mean bootwis (train)
# 18-23: test metrics
valid = ~np.isnan(recqvi[:, 0])
r = recqvi[valid]
print('models:', r.shape[0])
print('train bootql mean:', np.nanmean(r[:, 3]))
print('train bootwis mean:', np.nanmean(r[:, 5]))
print('test bootql mean:', np.nanmean(r[:, 19]))
print('test bootwis mean:', np.nanmean(r[:, 21]))
```

```python
# Check that policies look sane
print('OA unique actions:', np.unique(OA[~np.isnan(OA)]).astype(int)[:10], '...')
print('OA min/max:', np.nanmin(OA), np.nanmax(OA))
```
